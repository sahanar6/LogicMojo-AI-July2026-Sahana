```python
names = ["Ada", "Alan", "Grace", "Kay", "Katherine", "Ed"]
count = 0
for i, name in enumerate(names, start=1):
    if len(name) > 4:
        print(f"{i}. {name} (long)")
        count += 1
    else:
        print(f"{i}. {name}")

print(f"{count} of {len(names)} names are longer than four letters.")             
```

```python
readings = [12, -5, 0, 8, -3, 7]
count = 0
res = 0
for value in readings:
    if(value) > 0:
        print(f"{value}: positive")  
        count += 1
    elif(value) < 0:
        print(f"{value}: negative")
    else:
        print(f"{value}: Zero")
        
    res = res + value
    
print(f"Positives : {count}")  
print(f"total: {res}")      
```

```python
scores = [12, 47, 47, 8, 5, 47, 19]
highest = scores[0]
first_ind = 0
for i, value in enumerate(scores):
    if value > highest:
        highest = value
        first_ind = i

appear = 0
for value in scores:
    if value == highest:
        appear += 1

print("heighest:", highest) 
print("first seen at index" ,first_ind)
print("appears",appear, "times")            
```

```python
values = [3, 8, 8, 12, 5, 20]
is_sorted = True
first_drop = -1
for i in range(len(values) -1):
    if values[i] > values[i+1]:
        is_sorted = False
        if first_drop == -1:
            first_drop = i

print("sorted: ", is_sorted)
print(f"first drop at index {first_drop} ({values[first_drop]} then {values[first_drop + 1]})")            
```
