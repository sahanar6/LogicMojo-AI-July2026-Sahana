```python
import numpy as np
arr = np.array([2,4,6,8,10])
print("plus 10:", arr+10)
arr1 = np.array([1, 2, 3])
arr2 = np.array([4,5,6])
print("product:", arr1*arr2)
print("mean:", arr.mean())
```

```python
import numpy as np
data = np.array([3, 8, 1, 10, 6, 4])
mean = data.mean()
print("mean:", np.round(mean,2))
greater = data[data > mean]
print("greater:", greater)
print("count:", (data > mean).sum())
```

```python
import numpy as np 
qty = np.array([2, 1, 3])
price = np.array([10, 25, 5])
vectorized = np.sum(qty * price)
print("vectorized:", vectorized)
```

```python
import numpy as np 
scores = np.array([50, 20, 80, 40])
norm = (scores - scores.min()) / (scores.max() - scores.min())
print("normalized:",np.round(norm,2))
```

```python
import pandas as pd
df = pd.DataFrame({
    "name": ["Ada", "Alan", "Grace", "Kay"],
    "score": [91, 72, 85, 60],
    "team": ["A", "B", "A", "B"],
})
print(df)
print("scores:", list(df["score"]))
print(df[df["score"] >= 85])
```

```python
df["passed"] = df["score"]>=70
print(df)
print(df.groupby("team")["score"].mean())
```

```python
raw = pd.DataFrame({
    "city": ["A", "B", "C", "D"],
    "sales": [100.0, np.nan, 250.0, 80.0],
    "visits": [10, 5, 25, 8],
})
clean = raw.dropna().copy()
clean = clean[clean["sales"]>90].copy()
clean["per_visit"] = clean["sales"] / clean["visits"]
print(clean)
```

```python
sales = pd.DataFrame({
    "category": ["food", "toys", "food", "books", "toys"],
    "amount": [20, 15, 30, 10, 25],
})
total = sales.groupby("category") ["amount"].sum()
print(total)
print("top category:", total.idxmax(), "with", total.max())
```
